# Transpile unitary gates

This notebook lowers gates and circuits to the native `rz`, `sx`, `x`, and `cz` basis. The transpiler does not route qubits, schedule pulses, or optimize the resulting circuit.

```python
import numpy as np

from mb_client.circuit import QuantumCircuit, UnitaryGate
from mb_client.compilation import translate_basis
```

```python
%matplotlib inline
```

```python
circuit = QuantumCircuit(2, 2)
circuit.y(0)
circuit.h(1)
circuit.cx(0, 1)
circuit.barrier(0, 1)
circuit.measure(0, 0)
circuit.measure(1, 1)
circuit.draw(format='mpl')
```

```python
circuit_basis = translate_basis(circuit, basis_gates=["sx", "x", "rz", "cz"])
circuit_basis.draw(format='mpl')
```
